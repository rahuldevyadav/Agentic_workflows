```python
from dotenv import load_dotenv
import os

from anthropic import Anthropic

# Specify the path to your .env file
dotenv_path = "../.env"

load_dotenv(dotenv_path=dotenv_path)

client = Anthropic()
model = "claude-sonnet-4-6"
```

# context helper

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages):
    response = client.messages.create(
        model=model,
        messages=messages,
        max_tokens=1000,
        temperature=0.7,
    )
    return response.content[0].text
```

```python
# making a starting message
messages = []
# add in the initial user message
add_user_message(messages, "What is the capital of France?")
print(messages)
print("----")
# Pass the messages to the model and get a response
response = chat(messages)   
print(response)
print("----")

# add the model response to the messages list
add_assistant_message(messages, response)

# add a follow up user message
add_user_message(messages, "What is the population of that city?")
print(messages)
print("----")
# Pass the updated messages to the model and get a response
response = chat(messages)
print(response)
print("----")
```
