```python
from dotenv import load_dotenv
import os

from anthropic import Anthropic

dotenv_path = '../.env'
load_dotenv(dotenv_path)

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
def add_user_message(messages, content):
    messages.append({"role": "user", "content": content})

def add_assistant_message(messages, content):
    messages.append({"role": "assistant", "content": content})
    
def chat(messages, system_prompt = None):
    parameters = {
        "model": model,
        "messages": messages,
        "max_tokens": 1000,
    }
    if system_prompt is not None:
        parameters["system"] = system_prompt
    response = client.messages.create(**parameters)

    return response.content[0].text
```

```python
messages = []
system_prompt = """You are a patient math tutor. Do not directly answer a student's questions. Guide them to a solution step by step."""
while True:
    user_input = input("User:> ")
    add_user_message(messages, user_input)
 
    if user_input.lower() in ["exit", "quit"]:
        break
    
    assistant_response = chat(messages, system_prompt)
    print(f"Assistant: {assistant_response}")
    add_assistant_message(messages, assistant_response)
```

```python
def chatbot(system_prompt):
    messages = []
    while True:
        user_input = input("User:> ")
        add_user_message(messages, user_input)
        
        if user_input.lower() in ["exit", "quit"]:
            break
        
        assistant_response = chat(messages, system_prompt)
        print(f"Assistant: {assistant_response}")
        add_assistant_message(messages, assistant_response)
```
