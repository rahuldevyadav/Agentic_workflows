```python
from dotenv import load_dotenv
import os

from anthropic import Anthropic

dotenv_path = '../.env'
load_dotenv(dotenv_path)

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
def add_user_message(messages, content):
    messages.append({"role": "user", "content": content})

def add_assistant_message(messages, content):
    messages.append({"role": "assistant", "content": content})
    
# def chat(messages, system_prompt = None, temperature = 0.7):
#     parameters = {
#         "model": model,
#         "messages": messages,
#         "max_tokens": 1000,
#         "temperature": temperature
#     }
#     if system_prompt is not None:
#         parameters["system"] = system_prompt
#     response = client.messages.create(**parameters)

#     return response.content[0].text
```

```python
messages = []
add_user_message(messages, "What is 2 + 2?")

stream = client.messages.create(
    model=model,                    
    messages=messages, 
    system="You are a helpful assistant.",
    max_tokens=100,
    stream=True)

for chunk in stream:
    print(chunk)
```

```python
messages = []
add_user_message(messages, "Write a 10 line poem about a cat.")
with client.messages.stream(
    model=model,                    
    messages=messages, 
    system="You are a helpful assistant.",
    max_tokens=1000,
    temperature=0.7) as response:
    for chunk in response.text_stream:
        print(chunk, end="")
```
